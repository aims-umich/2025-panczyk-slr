```python
import numpy as np
import pandas as pd
from sklearn.metrics import cohen_kappa_score
```

```python
abstract_df = pd.read_excel("../../supplemental_materials/phase1_withN.xlsx", header=1)
fulltext_df = pd.read_excel("../../supplemental_materials/phase2_withN.xlsx", header=1)
```

```python
abstract_df.head()
```

```python
abstract_k = cohen_kappa_score(abstract_df["Majdi\'s Vote to Include"], abstract_df["Nataly\'s Vote to Include"])
print(f"Title/Abstract Cohen's Kappa: {abstract_k}")
```

```python
fulltext_df.head()
```

```python
full_k = cohen_kappa_score(fulltext_df["Majdi\'s Vote to Include"], fulltext_df["Nataly\'s Vote to Include"])
print(f"Full Text Cohen's Kappa: {full_k}")
```
