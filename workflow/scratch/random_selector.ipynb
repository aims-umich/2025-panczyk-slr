```python
import pandas as pd
import numpy as np
```

# Title and Abstract Screening Portion

```python
all_searches = pd.read_csv('2025-panczyk-slr-initial.csv')
print(len(all_searches))
```

```python
title_subset = all_searches.sample(60)
title_subset.columns
```

```python
title_subset = title_subset[["Author", "Publication Year", "Title", "DOI"]]
title_subset
```

```python
title_subset.to_csv('title_subset.csv')
```

# Full Text Screening Portion

```python
full_text_excluded = pd.read_csv('excluded.csv')
print(len(full_text_excluded))
full_text_ex_subset = full_text_excluded.sample(n=8)
full_text_ex_subset = full_text_ex_subset[["Authors", "Published Year", "Title", "DOI"]]
col_map = {
    "Authors": "Author",
    "Published Year": "Publication Year",
    "Title": "Title",
    "DOI": "DOI"
}
full_text_ex_subset.rename(columns=col_map, inplace=True)
full_text_ex_subset.head()
```

# QA Screening Portion

```python
final_included = pd.read_csv('2025-panczyk-slr-final.csv')
print(len(final_included))
```

```python
qa_subset = final_included.sample(n=7)
qa_subset = qa_subset[["Author", "Publication Year", "Title", "DOI"]]
qa_subset.to_csv('qa_subset.csv')
```

# Merging QA Included with Full Text Excluded

```python
full_text = pd.concat([qa_subset, full_text_ex_subset], ignore_index=True)
full_text_shuffled = full_text.sample(frac=1)
full_text_shuffled.to_csv('full_text_2.csv')
```
